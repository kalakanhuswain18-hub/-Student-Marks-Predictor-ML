```python
from sklearn.linear_model import LinearRegression 
import pandas as pd 
data=pd.read_csv("student_data_v3.csv")
data
```

```python
x=data[["Hours" ,"Revision"]]
y=data["Marks"]
model=LinearRegression()
model.fit(x,y)
hours=float(input("Enter study hours : "))
revision=float(input("Enter revision hours : "))
new_model=pd.DataFrame([[hours,revision]] , columns=["Hours" ,"Revision"])
pred=model.predict(new_model)
print(f"based on your {hours} hours of studing and {revision} hours of revision , you may score {pred[0]:.2f}")
```
